```python
# ============================================================
#              LIBRARY MANAGEMENT SYSTEM
# ============================================================

books = {
    1: {
        "title": "Python Programming",
        "author": "John Smith",
        "status": "Available"
    },
    2: {
        "title": "Data Science Basics",
        "author": "Robert Brown",
        "status": "Available"
    },
    3: {
        "title": "Machine Learning",
        "author": "Andrew Ng",
        "status": "Available"
    }
}


# ------------------------------------------------------------
# ADD BOOK
# ------------------------------------------------------------
def add_book():
    print("\n========== ADD BOOK ==========")

    title = input("Enter book title: ").strip()
    author = input("Enter author name: ").strip()

    if title == "" or author == "":
        print("Book title and author cannot be empty!")
        return

    new_id = max(books.keys(), default=0) + 1

    books[new_id] = {
        "title": title,
        "author": author,
        "status": "Available"
    }

    print("Book added successfully!")
    print("Book ID:", new_id)


# ------------------------------------------------------------
# VIEW BOOKS
# ------------------------------------------------------------
def view_books():
    print("\n========== ALL BOOKS ==========")

    if not books:
        print("No books available.")
        return

    for book_id, book in books.items():
        print("\nBook ID :", book_id)
        print("Title   :", book["title"])
        print("Author  :", book["author"])
        print("Status  :", book["status"])
        print("-------------------------------")


# ------------------------------------------------------------
# SEARCH BOOK
# ------------------------------------------------------------
def search_book():
    print("\n========== SEARCH BOOK ==========")

    keyword = input("Enter book title to search: ").strip().lower()

    found = False

    for book_id, book in books.items():
        if keyword in book["title"].lower():
            print("\nBook Found!")
            print("Book ID :", book_id)
            print("Title   :", book["title"])
            print("Author  :", book["author"])
            print("Status  :", book["status"])
            found = True

    if not found:
        print("Book not found.")


# ------------------------------------------------------------
# ISSUE BOOK
# ------------------------------------------------------------
def issue_book():
    print("\n========== ISSUE BOOK ==========")

    try:
        book_id = int(input("Enter Book ID: "))
    except ValueError:
        print("Please enter a valid number!")
        return

    if book_id not in books:
        print("Book ID not found.")
        return

    if books[book_id]["status"] == "Issued":
        print("This book is already issued.")
        return

    books[book_id]["status"] = "Issued"

    print("Book issued successfully!")
    print("Book:", books[book_id]["title"])


# ------------------------------------------------------------
# RETURN BOOK
# ------------------------------------------------------------
def return_book():
    print("\n========== RETURN BOOK ==========")

    try:
        book_id = int(input("Enter Book ID: "))
    except ValueError:
        print("Please enter a valid number!")
        return

    if book_id not in books:
        print("Book ID not found.")
        return

    if books[book_id]["status"] == "Available":
        print("This book has not been issued.")
        return

    books[book_id]["status"] = "Available"

    print("Book returned successfully!")
    print("Book:", books[book_id]["title"])


# ------------------------------------------------------------
# DELETE BOOK
# ------------------------------------------------------------
def delete_book():
    print("\n========== DELETE BOOK ==========")

    try:
        book_id = int(input("Enter Book ID: "))
    except ValueError:
        print("Please enter a valid number!")
        return

    if book_id not in books:
        print("Book ID not found.")
        return

    if books[book_id]["status"] == "Issued":
        print("Issued book cannot be deleted.")
        return

    deleted_book = books.pop(book_id)

    print("Book deleted successfully!")
    print("Deleted book:", deleted_book["title"])


# ------------------------------------------------------------
# MAIN MENU
# ------------------------------------------------------------
def main():

    while True:

        print("\n========================================")
        print("       LIBRARY MANAGEMENT SYSTEM")
        print("========================================")
        print("1. Add Book")
        print("2. View All Books")
        print("3. Search Book")
        print("4. Issue Book")
        print("5. Return Book")
        print("6. Delete Book")
        print("7. Exit")
        print("========================================")

        choice = input("Enter your choice (1-7): ").strip()

        if choice == "1":
            add_book()

        elif choice == "2":
            view_books()

        elif choice == "3":
            search_book()

        elif choice == "4":
            issue_book()

        elif choice == "5":
            return_book()

        elif choice == "6":
            delete_book()

        elif choice == "7":
            print("\nThank you for using Library Management System!")
            break

        else:
            print("Invalid choice! Please enter a number from 1 to 7.")


# ------------------------------------------------------------
# START PROGRAM
# ------------------------------------------------------------

main()
```
